# Recitation 1: Bond Prices and Yields

This recitation reviews how to value bonds, how to interpret the different yield measures, and how the term structure of interest rates works.


## 1. Accrued Interest and Quoted Bond Price

When a bond is traded between two coupon dates, the seller has already earned part of the upcoming coupon. The buyer compensates the seller for this portion — the **accrued interest**.

* **Clean Price (Quoted Price or Flat Price):** Price **excluding** accrued interest. This is the price you see on trading screens and in financial news.
* **Dirty Price (Full Price / Invoice Price):** Price the buyer **actually pays**.
* **Accrued Interest (AI):** Interest earned by the seller since the last coupon but not yet paid.

$$
\text{Accrued Interest} = \text{Coupon Payment} \times \frac{\text{Days since last coupon}}{\text{Days in coupon period}}
$$

$$
\text{Full (Dirty) Price} = \text{Flat (Clean) Price} + \text{Accrued Interest}
$$

#### The sawtooth pattern
* As time passes between coupon payments, interest accrues daily, causing the **dirty price** to rise gradually.
* The **clean price** moves smoothly with market yields but excludes accrued interest.
* When the coupon is paid, accrued interest resets to zero, creating a **saw-tooth pattern** in the dirty price.

### 1. Câu chuyện "Góp vốn mở tiệm Trà sữa" (Ví dụ thực tế)

Giả sử bạn phát hành một **"Trái phiếu Trà sữa"** cho đứa bạn cùng lớp với các điều kiện sau:

* **Mệnh giá (Par Value):** 1.000.000 VNĐ.
* **Tiền lãi (Coupon):** 120.000 VNĐ/năm (mỗi năm trả 1 lần vào **ngày 31/12**).
* **Chu kỳ:** Bạn trả đúng 120k tiền lãi cho người nào **đang cầm tờ trái phiếu vào ngày 31/12**.

Bây giờ hãy xem chuyện gì xảy ra vào ngày **30/06** (đúng giữa năm):

* Người bạn A cầm tờ trái phiếu này từ ngày 01/01 đến 30/06 (được **6 tháng**).
* Do cần tiền đi du lịch, bạn A muốn bán lại trái phiếu này cho bạn B vào ngày 30/06.
* Giả sử giá trị gốc của tờ trái phiếu trên thị trường lúc này không đổi, vẫn là **1.000.000 VNĐ**.

#### Chuyện gì xảy ra nếu bạn B chỉ trả cho bạn A đúng 1.000.000 VNĐ?

* Bạn A đã "gánh" nợ và cho bạn vay trong 6 tháng đầu năm, nhưng **không nhận được xu tiền lãi nào**.
* Bạn B vừa mới mua tờ trái phiếu vào giữa năm, cầm đúng 6 tháng còn lại, nhưng đến ngày 31/12 lại được bạn **trả trọn vẹn 120.000 VNĐ tiền lãi cho cả năm**.
👉 **Như vậy là bất công cho bạn A!** 6 tháng tiền lãi (tương đương 60.000 VNĐ) vốn dĩ thuộc về công sức của bạn A.

### 2. Giải pháp tài chính: Clean Price vs. Dirty Price

Để công bằng, khi bạn B mua lại trái phiếu từ bạn A vào ngày 30/06, **bạn B phải bù cho bạn A phần tiền lãi mà bạn A đã tích lũy được trong 6 tháng qua**.

Từ đây, dân Tài chính (BFIN) chia ra 3 thuật ngữ cực kỳ rõ ràng:

1. **Accrued Interest (AI - Lãi dồn tích / Lãi tích lũy):**
* Là khoản tiền lãi đã phát sinh kể từ ngày trả lãi gần nhất nhưng chưa được thanh toán.
* **Công thức tính:** $\text{Coupon} \times \frac{\text{Số ngày đã qua}}{\text{Số ngày trong kỳ}}$
* *Ở ví dụ trên:* $120.000 \times \frac{6 \text{ tháng}}{12 \text{ tháng}} = 60.000 \text{ VNĐ}$.


2. **Clean Price / Quoted Price / Flat Price (Giá sạch / Giá niêm yết):**
* Là giá trị thuần của trái phiếu **CHƯA** tính phần lãi dồn tích. Đây là mức giá bạn thấy hiển thị trên bảng điện tử hoặc màn hình giao dịch (như Bloomberg hay giao dịch chứng khoán).
* *Ở ví dụ trên:* Clean Price = **1.000.000 VNĐ**.


3. **Dirty Price / Full Price / Invoice Price (Giá bẩn / Giá thanh toán thực tế):**
* Là tổng số tiền thực tế mà người mua (bạn B) **phải rút ví trả** cho người bán (bạn A).
* **Công thức:** $\text{Dirty Price} = \text{Clean Price} + \text{Accrued Interest}$
* *Ở ví dụ trên:* $\text{Dirty Price} = 1.000.000 + 60.000 = \mathbf{1.060.000 \text{ VNĐ}}$.



---

### 3. Tại sao lại cần phân biệt "Clean Price" và "Dirty Price"?

* **Tại sao sàn giao dịch chỉ niêm yết Clean Price?**
Nếu niêm yết Dirty Price, giá trái phiếu trên màn hình sẽ **tăng dần mỗi ngày** (do lãi tích lũy tăng lên hàng ngày) rồi đột ngột **rớt thảm hại về 0** ngay sau ngày trả lãi. Việc biến động nhảy vọt này làm nhà đầu tư khó nhận biết được giá trị thực sự của trái phiếu đang tăng hay giảm do lãi suất thị trường thay đổi.
Vì vậy, người ta dùng **Clean Price** để đường biểu diễn giá trông phẳng và phản ánh đúng xu hướng thị trường.
* **Hình ảnh "Răng cưa" (Sawtooth Pattern):**
* **Clean Price:** Chạy tương đối êm đềm, phẳng lỳ.
* **Dirty Price:** Nhìn giống như **lưỡi cưa** — tăng dần đều từng ngày theo thời gian tích lãi, đến ngày nhận lãi coupon xong thì "sập" xuống về đúng bằng Clean Price, rồi lại từ từ tăng lên cho kỳ tiếp theo.

## 2. Option-Embedded and Featured Bonds

$$
\text{Price (callable)} < \text{Price (straight)} < \text{Price (puttable)}
$$

* **Callable bonds:** the issuer can redeem early $\implies$ **call risk** for the investor.
* **Puttable bonds:** the investor can sell the bond back at par.
* **Convertible bonds:** carry an equity option (conversion ratio, conversion premium).
* **Floating-rate notes (FRN):** coupon resets to a reference rate plus a spread.

## 3. Conventional Yield Measures

Several yield measures are commonly quoted by dealers and used by portfolio managers: **current yield**, **yield to maturity**, **yield to call**, yield to refunding, yield to put and cash flow yield.

### 3.1. Current Yield
The annual coupon interest relative to the market price. It **ignores capital gains and the time value of money**.

$$
\text{Current yield} = \frac{\text{annual dollar coupon}}{\text{price}}
$$

* **Example:** A 15-year 7% p.a. US Treasury bond, par value \$1,000, is selling for \$769.40. What is the current yield of the bond?

### 3.2. Yield to Maturity (YTM)
* YTM is the **IRR** of a bond's cash flows.
* YTM reflects the **total return** (coupons + capital gain/loss) if the bond is held to maturity.

$$
P = \frac{c}{1+y} + \frac{c}{(1+y)^2} + \cdots + \frac{c}{(1+y)^n} + \frac{FV}{(1+y)^n}
  = \sum_{t=1}^{n} \frac{c}{(1+y)^t} + \frac{FV}{(1+y)^n}
$$

Where:
* $P$ = price (in dollars)
* $n$ = number of periods
* $c$ = coupon payment per period (in dollars)
* $y$ = periodic interest rate (also written $i$ or $r$)
* $FV$ = face value
* $t$ = time period when the payment is received

### 3.3. Coupon Rate – Current Yield – YTM Relationship

| Bond selling at: | Relationship |
| :---: | :--- |
| **Par** | Coupon rate = current yield = yield to maturity |
| **Discount** | Coupon rate < current yield < yield to maturity |
| **Premium** | Coupon rate > current yield > yield to maturity |

> Premium bonds have a high coupon $\implies$ lower return (YTM) because the price declines toward par over time.

<center>

</center>

### 3.4. Yield to Call
If the bond has a **call provision**, the issuer can call the bond before maturity at the call price (or the appropriate price in the call schedule).

$$
P = \sum_{t=1}^{n^*} \frac{c}{(1+y)^t} + \frac{\text{Call Price}}{(1+y)^{n^*}}
$$

Where:
* Call Price (in dollars)
* $n^*$ = number of periods until the assumed call date

### 1. Current Yield (Lợi suất hiện hành) — *Cực kỳ đơn giản*

#### 💡 Ví dụ tiệm trà sữa:

* Tờ trái phiếu có **mệnh giá 1.000.000 VNĐ**, tiền lãi cố định mỗi năm bạn nhận được là **70.000 VNĐ** (Coupon rate = 7%).
* Nhưng hiện tại, tiệm trà sữa kinh doanh hơi chậm nên bạn mua lại tờ trái phiếu này trên "thị trường sinh viên" với giá rẻ hơn: **769.400 VNĐ** (gọi là mua với giá chiết khấu - Discount).

**Current Yield** trả lời cho câu hỏi: *"Năm nay tui bỏ ra 769.400 VNĐ để mua, tui đút túi 70.000 VNĐ tiền lãi thì tỷ lệ sinh lời ngay trước mắt là bao nhiêu?"*

$$\text{Current Yield} = \frac{\text{Tiền lãi nhận được hàng năm}}{\text{Giá mua thực tế trên thị trường}} = \frac{70.000}{769.400} \approx \mathbf{9{,}10\%}$$

* **Ưu điểm:** Tính siêu nhanh, dễ hiểu.
* **Nhược điểm:** **Bỏ qua yếu tố thời gian** và **bỏ qua khoản chênh lệch giá** khi đáo hạn (vì khi hết hạn bạn nhận lại tận 1.000.000 VNĐ gốc, lời thêm một khoản chênh lệch giá nữa nhưng Current Yield chưa tính vào).

---

### 2. Yield to Maturity (YTM - Lợi suất đến ngày đáo hạn) — *Thước đo chuẩn chỉnh nhất*

#### 💡 Ví dụ tiệm trà sữa:

Giả sử tờ trái phiếu trà sữa giá 769.400 VNĐ ở trên còn tới **15 năm** mới đáo hạn.
Nếu bạn giữ tờ trái phiếu này **đúng 15 năm nữa cho tới ngày hết hạn**, dòng tiền bạn nhận được gồm:

1. Mỗi năm nhận **70.000 VNĐ** tiền lãi coupon (liên tục 15 năm).
2. Đến năm thứ 15, nhận lại **1.000.000 VNĐ** tiền gốc ban đầu.

**YTM** chính là **tỷ lệ sinh lời thực tế trung bình mỗi năm (IRR)** của toàn bộ quá trình đó, bao gồm cả tiền lãi coupon lẫn khoản lời chênh lệch từ giá mua $769.400 \to 1.000.000 \text{ VNĐ}$.

$$\text{Giá trái phiếu } (P) = \sum_{t=1}^{n} \frac{c}{(1+y)^t} + \frac{FV}{(1+y)^n}$$

*(Trong đó $y$ chính là YTM. Để tính $y$, dân BBA thường dùng máy tính tài chính BA II Plus, Excel hoặc hàm Python).*

### 3. Mối quan hệ giữa Coupon Rate – Current Yield – YTM

Để đi thi trắc nghiệm không bị lừa, bạn chỉ cần nhớ quy tắc bàn cân giá trái phiếu:

1. **Bán đúng giá niêm yết (Trading at Par - Giá 1.000.000 VNĐ):**

$$\text{Coupon Rate} = \text{Current Yield} = \text{YTM}$$



*(Mua đúng giá gốc thì lãi suất ghi trên tờ giấy đúng bằng lãi suất bạn hưởng).*
2. **Bán giảm giá / Chiết khấu (Trading at Discount - Giá 769.400 VNĐ):**

$$\text{Coupon Rate} < \text{Current Yield} < \text{YTM}$$



*Vì bạn mua rẻ hơn mệnh giá, ngoài tiền lãi coupon bạn còn lời thêm khoản chênh lệch khi nhận lại đủ 1.000.000 VNĐ lúc đáo hạn, nên tổng lợi suất YTM sẽ là cao nhất.*
3. **Bán tăng giá / Thượng giá (Trading at Premium - Ví dụ giá 1.200.000 VNĐ):**

$$\text{Coupon Rate} > \text{Current Yield} > \text{YTM}$$



*Vì tiệm trà sữa quá đắt hàng, bạn phải mua đắt hơn mệnh giá. Đến ngày đáo hạn chỉ nhận lại 1.000.000 VNĐ (bị lỗ phần chênh lệch giá mua), nên tổng lợi suất YTM thực nhận sẽ bị kéo xuống thấp nhất.*

---

### 4. Yield to Call (YTC - Lợi suất đến ngày bị thu hồi)

#### 💡 Ví dụ tiệm trà sữa:

Giả sử khi phát hành trái phiếu, tiệm trà sữa có gài thêm điều khoản **Callable**: *"Tụi tui có quyền mua lại tờ trái phiếu này sớm ở năm thứ 5 với giá 1.050.000 VNĐ nếu tiệm làm ăn quá lời và muốn trả nợ sớm."*

Nếu tiệm trà sữa thu hồi lại trái phiếu ở năm thứ 5:

* Bạn chỉ nhận coupon trong **5 năm** (thay vì 15 năm).
* Đến năm thứ 5 bạn nhận lại **1.050.000 VNĐ** (Call Price) chứ không đợi đến 1.000.000 VNĐ ở năm 15.

👉 **Yield to Call (YTC)** chính là YTM được tính dựa trên mốc thời gian rút ngắn này (5 năm) và số tiền nhận lại lúc thu hồi ($1.050.000$).

## 4. The Term Structure of Interest Rates

* The term structure of interest rates shows the relationship between **bond yields** and **time to maturity**.
* It is usually illustrated by a **yield curve**.
* **1-year forward rate starting at year $t$:**

$$
(1 + S_{t+1})^{t+1} = (1 + S_t)^{t} (1 + f_{t,t+1})
$$

* Solve for the forward rate:

$$
f_{t,t+1} = \frac{(1 + S_{t+1})^{t+1}}{(1 + S_t)^{t}} - 1
$$

### 1. Ý tưởng cốt lõi: Vay ngắn hạn vs. Vay dài hạn

Hãy tưởng tượng bạn và nhóm bạn kinh doanh trà sữa muốn huy động vốn từ sinh viên trong trường:

* **Kịch bản A:** Bạn mượn tiền bạn bè trong **1 năm**.
* **Kịch bản B:** Bạn mượn tiền bạn bè trong **10 năm**.

**Hỏi:** Ở kịch bản nào thì các bạn sinh viên sẽ đòi hỏi **lãi suất cao hơn**?

Chắc chắn là **10 năm**! Vì cho vay trong 10 năm gặp vô vàn rủi ro: tiệm trà sữa có thể bị giải thể, lạm phát khiến tiền mất giá, hoặc sinh viên đó ra trường không biết thu hồi nợ thế nào.
👉 Do đó, **thời hạn vay càng dài thì người cho vay càng đòi lãi suất cao hơn để bù đắp rủi ro**.

**Cấu trúc kỳ hạn (Term Structure)** chính là mối quan hệ giữa **thời gian cho vay (Maturity)** và **lãi suất/lợi suất tương ứng (Yield)** tại cùng một thời điểm. Khi vẽ mối quan hệ này lên đồ thị, bạn sẽ thu được **Đường cong lợi suất (Yield Curve)**.

### 2. Spot Rate ($S_t$) vs. Forward Rate ($f_{t, t+1}$)

Để làm bài tập phần này, dân BBA chỉ cần phân biệt 2 loại lãi suất sau:

#### A. Spot Rate ($S_t$ - Lãi suất điểm / Lãi suất giao ngay)

* Là lãi suất cho một khoản vay **bắt đầu ngay hôm nay** và kéo dài trong $t$ năm.
* **Ví dụ trà sữa:**
* Bạn mượn tiền **ngay bây giờ cho 1 năm**: Lãi suất $S_1 = 5\%$/năm.
* Bạn mượn tiền **ngay bây giờ cho 2 năm**: Lãi suất $S_2 = 6\%$/năm.



#### B. Forward Rate ($f_{t, t+1}$ - Lãi suất kỳ hạn)

* Là lãi suất cho một khoản vay trong tương lai (ví dụ: **bắt đầu từ năm thứ $t$ và kéo dài đến năm thứ $t+1$**), nhưng **thỏa thuận mức lãi suất đó ngay từ hôm nay**.
* **Ví dụ trà sữa:** Bạn dự định năm sau mới mở thêm chi nhánh thứ 2. Bạn thỏa thuận trước với bạn A: *"Năm sau tui mượn ông tiền trong vòng 1 năm (từ năm 1 đến năm 2), tính lãi bao nhiêu?"* 👉 Lãi suất thỏa thuận trước này chính là **Forward Rate $f_{1, 2}$**.

---

### 3. Nguyên lý Không kinh doanh chênh lệch giá (No-Arbitrage Principle)

Làm sao để tính ra con số $f_{1, 2}$ hợp lý? Hãy đặt 2 lựa chọn đầu tư 2 năm lên bàn cân:

* **Lựa chọn 1:** Mang 100k đi gửi/cho vay 1 mạch **2 năm liền** với lãi suất $S_2 = 6\%$/năm.

$$\text{Số tiền nhận được sau 2 năm} = 100 \times (1 + S_2)^2 = 100 \times (1 + 0{,}06)^2$$


* **Lựa chọn 2:** Mang 100k cho vay **năm thứ 1** với lãi suất $S_1 = 5\%$. Sau đó lấy toàn bộ gốc + lãi tiếp tục cho vay ở **năm thứ 2** với lãi suất kỳ hạn $f_{1, 2}$.

$$\text{Số tiền nhận được sau 2 năm} = 100 \times (1 + S_1)^1 \times (1 + f_{1, 2})^1$$



Để thị trường cân bằng (không ai ăn chênh lệch giá), kết quả của 2 lựa chọn này **phải bằng nhau**:

$$(1 + S_2)^2 = (1 + S_1)^1 \times (1 + f_{1, 2})^1$$

---

### 4. Công thức tổng quát & Cách tính Forward Rate

Từ nguyên lý trên, ta có công thức tổng quát tính Forward Rate 1 năm bắt đầu từ năm $t$:

$$(1 + S_{t+1})^{t+1} = (1 + S_t)^t \times (1 + f_{t, t+1})$$

$$\Rightarrow f_{t, t+1} = \frac{(1 + S_{t+1})^{t+1}}{(1 + S_t)^t} - 1$$

#### 💡 Ví dụ tính toán thực tế cho bài thi BBA:

Cho $S_1 = 5\%$ và $S_2 = 6\%$. Tính lãi suất kỳ hạn $f_{1, 2}$ (lãi suất 1 năm bắt đầu từ năm 1)?

$$f_{1, 2} = \frac{(1 + 0{,}06)^2}{(1 + 0{,}05)^1} - 1 = \frac{1{,}1236}{1{,}05} - 1 \approx 7{,}01\%$$

**Ý nghĩa:** Để việc gửi tiền từng năm một đạt lợi nhuận ngang bằng với việc gửi liền 2 năm (6%/năm), thì lãi suất của năm thứ 2 ($f_{1, 2}$) phải đạt mức **7.01%**.

---

### 💡 Bảng tóm tắt nhanh cho dân BBA

| Thuật ngữ | Ký hiệu | Ý nghĩa trong ví dụ tiệm trà sữa |
| --- | --- | --- |
| **Spot Rate** | $S_t$ | Lãi suất mượn tiền **từ hôm nay** đến năm $t$. |
| **Forward Rate** | $f_{t, t+1}$ | Lãi suất chốt trước hôm nay cho khoản vay **từ năm $t$ đến năm $t+1$**. |
| **Yield Curve** | Đồ thị $(t, S_t)$ | Đường biểu diễn lãi suất theo thời gian (thường đi lên do vay dài hạn rủi ro hơn). |

Có nghĩa là công thức này nó trả lời cho câu hỏi là:
Nếu tôi đầu tư 6% cho xuyên suốt từ đầu năm 1 (cuối mốc 0) đến đầu năm 3 (cuối năm 2) thì tôi nhận được từng này, vậy thì nếu tôi chơi bài là sau hết năm thứ nhất, tôi dùng cả tiền gốc lẫn lãi trong vòng 1 năm đi đầu tư tiếp thêm 1 năm, thì cái phần trăm lãi suất của cái năm này nó phải là bao nhiêu để bằng được cái số tiền tôi dùng cả 2 năm xuyên suốt để đầu tư đúng không


**Chính xác 100%!** Tư duy của bạn hoàn toàn chuẩn xác và chạm đúng bản chất của nguyên lý **Không kinh doanh chênh lệch giá (No-Arbitrage Principle)** trong tài chính.

Để khẳng định lại logic của bạn một cách rõ ràng nhất:

### Lựa chọn 1: Đầu tư "xuyên suốt" 2 năm (Gửi 1 mạch)

* Bạn mang tiền đi gửi ngay từ **cuối mốc 0** kéo dài đến **cuối năm 2** (tròn 2 năm).
* Lãi suất $S_2 = 6\%$/năm.
* **Số tiền thu về:** $(1 + 0{,}06)^2$

### Lựa chọn 2: "Cuốn chiếu" từng năm một (Reinvest)

* **Năm 1:** Gửi 1 năm với lãi suất $S_1 = 5\%$. Cuối năm 1 bạn thu về cả gốc lẫn lãi là $(1 + 0{,}05)^1$.
* **Năm 2:** Lấy **toàn bộ cục tiền (gốc + lãi) đó** mang đi đầu tư tiếp thêm 1 năm nữa với lãi suất $f_{1, 2}$.
* **Số tiền thu về:** $(1 + 0{,}05)^1 \times (1 + f_{1, 2})^1$

---

### Câu hỏi cốt lõi mà công thức $f_{1, 2}$ trả lời:

Mức lãi suất $f_{1, 2}$ ở năm thứ 2 **phải bằng bao nhiêu** để số tiền thu về từ **Lựa chọn 2** cày cuốc đúng bằng số tiền thu về từ **Lựa chọn 1**?

$$(1 + 0{,}05)^1 \times (1 + f_{1, 2})^1 = (1 + 0{,}06)^2$$

$$f_{1, 2} = \frac{(1 + 0{,}06)^2}{(1 + 0{,}05)^1} - 1 \approx 7{,}01\%$$

👉 Như vậy, nếu lãi suất năm thứ 2 đạt **7.01%**, thì việc bạn "chơi bài cuốn chiếu từng năm" hay "gửi một mạch 2 năm" sẽ cho ra kết quả **về tiền là hệt như nhau**!

Việc tính toán **Forward Rate ($f_{t, t+1}$)** không chỉ là một công thức toán học suông để đi thi BBA, mà nó giải quyết các bài toán chiến lược cực kỳ lớn trên thực tế.

Dưới góc nhìn tài chính thực tế, Forward Rate đại diện cho **"Mức lãi suất kỳ vọng của thị trường trong tương lai"**.

Dưới đây là mục đích thực tế của nó ở cả hai góc nhìn: **Người đi vay (Tiệm Trà Sữa)** và **Nhà đầu tư (Sinh viên cho vay)**.

### 1. Góc nhìn Người đi vay (Tiệm Trà Sữa) — Bài toán: *Nên chốt lãi suất vay từ bây giờ hay chờ đến năm sau mới vay?*

Giả sử tiệm trà sữa của bạn đang lên kế hoạch mở rộng chi nhánh 2 vào **năm sau**. Bạn chắc chắn sẽ cần vay **100 triệu VNĐ** trong vòng 1 năm (từ năm 1 đến năm 2).

Hiện tại, qua tính toán từ Yield Curve, bạn biết Forward Rate cho năm sau là **$f_{1,2} = 7{,}01\%$**.

* **Bài toán thực tế:**
Ngân hàng/Thị trường sẵn sàng cho bạn ký hợp đồng **chốt trước lãi suất** vay cho năm sau ở mức **$6{,}8\%$**. Bạn có nên ký không?
* **Cách ra quyết định:**
* Thị trường đang kỳ vọng lãi suất năm sau là **$7{,}01\%$**.
* Ngân hàng cho bạn chốt trước ở mức **$6{,}8\%$** (thấp hơn kỳ vọng thị trường).
👉 **Quyết định:** Bạn nên **ký hợp đồng chốt trước ngay hôm nay (Hedging / Khóa rủi ro lãi suất)**. Nhờ đó, dù năm sau thị trường có biến động kinh hoàng hay lãi suất tăng vọt, tiệm trà sữa của bạn vẫn an tâm vay được tiền với giá rẻ $6{,}8\%$.

### 2. Góc nhìn Nhà đầu tư (Sinh viên có tiền) — Bài toán: *Nên gửi tiền dài hạn ngay hay gửi ngắn hạn rồi chờ cơ hội?*

Bạn có **100 triệu VNĐ** tiền nhàn rỗi trong 2 năm.

* **Bài toán thực tế:**
Bạn đứng trước 2 lựa chọn:
1. Gửi tiết kiệm cố định **2 năm liền** với lãi suất $S_2 = 6\%$/năm.
2. Gửi **1 năm** với lãi suất $S_1 = 5\%$, rồi hết năm thứ nhất mới mang cả gốc lẫn lãi đi gửi tiếp năm thứ 2.


* **Cách ra quyết định:**
Bạn tính ra $f_{1,2} = 7{,}01\%$. Con số này trở thành **mức lãi suất mốc (Benchmark)** để bạn so sánh:
* **Nếu bạn dự báo:** Năm sau nền kinh tế sẽ lạm phát, Ngân hàng Nhà nước sẽ siết tiền tệ và lãi suất thực tế năm sau sẽ vọt lên **$8\%$** (cao hơn mốc $7{,}01\%$):
👉 **Quyết định:** Bạn **KHÔNG** gửi 2 năm liền. Bạn chọn gửi 1 năm ($5\%$), chờ sang năm nhận lãi $8\%$ để thu về tổng lợi nhuận cao hơn.
* **Nếu bạn dự báo:** Năm sau kinh tế suy thoái, lãi suất sẽ hạ về **$5{,}5\%$** (thấp hơn mốc $7{,}01\%$):
👉 **Quyết định:** Bạn **chốt ngay** gói gửi 2 năm liền với lãi suất $6\%$/năm để "khóa" lợi nhuận cao trước khi lãi suất thị trường giảm thảm hại.

| **Nhà đầu tư / Quỹ** | **Định giá rủi ro & Tìm cơ hội đầu tư (Trading Strategy)** | So sánh nhận định cá nhân với con số kỳ vọng của thị trường ($f$) để chọn chiến lược gửi tiền/mua trái phiếu tối ưu. |

Nếu đúng như bạn nói — năm nay bạn **chưa hề có kế hoạch** mở chi nhánh 2, chỉ vay 100 triệu làm chi nhánh 1. Sang năm thứ 2, thấy làm ăn xịn quá đột ngột nảy ra ý định mở thêm chi nhánh 2:

### Tình huống 2: Quyết định CÓ KẾ HOẠCH TỪ TRƯỚC (Nơi Benchmark ra đời)

Trong quản trị tài chính doanh nghiệp, đa số các dự án mở rộng đều nằm trong **Kế hoạch phát triển (Business Plan)** 3–5 năm.

Giả sử ngay từ đầu năm 1, bạn đã chốt kế hoạch:

Lúc này, câu hỏi đặt ra ở mốc Năm 0 là: **Thị trường lấy đâu ra con số $f_{1,2} = 7{,}01\%$ để làm Benchmark cho bạn?**

#### Con số Benchmark $7{,}01\%$ từ đâu mà có?

Nó **không phải do bạn tự nghĩ ra**, cũng **không phải do ngân hàng phán bừa**, mà nó được trích xuất trực tiếp từ **Đường cong lợi suất (Yield Curve)** trên thị trường tài chính ngay tại thời điểm hôm nay:

1. Bạn nhìn lên bảng điện tử ngân hàng/trái phiếu chính phủ:
* Lãi suất vay 1 năm ($S_1$) đang là **5%**.
* Lãi suất vay 2 năm ($S_2$) đang là **6%**.


2. Áp dụng công thức No-Arbitrage:

$$f_{1,2} = \frac{(1 + 0{,}06)^2}{(1 + 0{,}05)^1} - 1 = \mathbf{7{,}01\%}$$


3. Con số **$7{,}01\%$** này chính là **"mức giá hòa vốn/kỳ vọng chung của toàn thị trường"** cho khoản vay 1 năm vào năm sau.

### Kết hợp lại: Bài toán so sánh với Benchmark diễn ra như thế nào?

Ngay tại Năm 0, khi bạn biết Benchmark thị trường là $7{,}01\%$, bạn có 2 lựa chọn tài chính:

* **Lựa chọn A (Thả nổi - Chờ năm sau vay):**
Bạn không làm gì cả. Năm sau đến ngân hàng vay theo lãi suất thực tế lúc đó. Bạn đang đánh cược: nếu năm sau kinh tế hạ nhiệt, lãi suất xuống $6\%$ thì bạn lời; nếu lãi suất vọt lên $8\%$ thì bạn lỗ so với mốc $7{,}01\%$.
* **Lựa chọn B (Khóa lãi suất trước - Hedging / Forward Contract):**
Bạn đến ngân hàng hoặc quỹ đầu tư thương lượng: *"Tôi muốn ký hợp đồng vay trước cho năm sau"*.
Ngân hàng nhìn vào Benchmark thị trường $7{,}01\%$ và bảo: *"Ok, tui cho bạn chốt cứng lãi suất vay cho năm sau là $6{,}8\%$."*
👉 Vì $6{,}8\% < 7{,}01\%$ (Benchmark), bạn biết ngay đây là **mức giá vay có lời**, bạn ký chốt ngay để triệt tiêu toàn bộ rủi ro biến động lãi suất năm sau!

---

### 💡 Tóm lại cho dễ nhớ:

* Nếu là **quyết định bất chợt**, bạn là "người đi chợ mua hàng theo giá niêm yết tại thời điểm đó", không dùng Forward Rate để khóa giá được nữa.
* Nếu là **dự án có lộ trình**, Forward Rate ($f_{t, t+1}$) chính là **cây thước đo (Benchmark)** do thị trường cung cấp sẵn hôm nay, giúp bạn quyết định nên **"chốt giá vay ngay từ bây giờ"** hay **"để năm sau vay thả nổi"**.

```python
import numpy as np


def bisection_root(function, low=1e-8, high=1.0, tolerance=1e-12, max_iterations=200):
    """Find a root with bisection; no SciPy installation is required."""
    f_low, f_high = function(low), function(high)
    if f_low * f_high > 0:
        raise ValueError("The root is not bracketed by the chosen interval.")
    for _ in range(max_iterations):
        midpoint = (low + high) / 2
        f_midpoint = function(midpoint)
        if abs(f_midpoint) < tolerance:
            return midpoint
        if f_low * f_midpoint < 0:
            high, f_high = midpoint, f_midpoint
        else:
            low, f_low = midpoint, f_midpoint
    return (low + high) / 2


def bond_price(coupon_rate, ytm, years, freq=2, face=1000):
    """Price of a plain-vanilla bond on a coupon date."""
    c = coupon_rate * face / freq
    y = ytm / freq
    n = int(round(years * freq))
    return c * (1 - (1 + y) ** -n) / y + face / (1 + y) ** n

def bond_ytm(price, coupon_rate, years, freq=2, face=1000):
    """Annualised YTM (APR) that equates the PV of cash flows to the price."""
    f = lambda y: bond_price(coupon_rate, y, years, freq, face) - price
    return bisection_root(f, 1e-6, 1.0)

def clean_dirty(coupon_rate, ytm, years, days_since, days_in_period, freq=2, face=1000):
    """Full (dirty) and flat (clean) price when settling between coupon dates."""
    p0 = bond_price(coupon_rate, ytm, years, freq, face)          # price at last coupon date
    w = days_since / days_in_period
    full = p0 * (1 + ytm / freq) ** w                              # accrete forward to settlement
    ai = (coupon_rate * face / freq) * w
    return full, ai, full - ai

# --- Example: current yield of a 15-year 7% Treasury selling for $769.40 ---
print("Example: current yield of a 15-year 7% Treasury selling for $769.40")
print(f"Current yield = {70 / 769.40:.2%}")
```

| :--- | :--- |
| Par Value | \$1,000 |
| Annual Coupon Rate | 9% (paid semiannually) |
| Maturity | 20 years |
| Callable | In 5 years at 105% of par (\$1,050) |
| Current Yield | 8.18% |

* Calculate the current price of the bond based on the current yield.
* Calculate the yield to call (YTC), assuming the bond will be called in 5 years.
* Interpret whether investors are likely to expect the bond to be called.

Từ từ nhé, cái yield tức là cái lãi suất tăng trưởng hàng năm, với mỗi loại Discount hay Premium Bond thì nó sẽ tăng dần hoặc giảm dần về phía YTM hoặc YTC tùy thời điểm và trường hợp xảy ra, chuẩn chưa?

Và việc đánh giá xem liệu cái Bond nó có thể bị Call hay không tại thời điểm t cho trước, nó sẽ chả liên quan gì đến YTC cả đúng không, bởi vì bản chất đang tính YTC là: "Tui lời được bao nhiêu, nếu bạn mua lại Bond". Còn việc đánh giá nó mua lại không thì đề phải cho cái Interest Rate của thị trường, và mình phải so sánh với cái Coupon Rate định sẵn cố định của Bond chứ đúng không

1. Về đường đi của Yield và Giá trái phiếu (Pull-to-Par Effect)
Ý bạn hiểu hoàn toàn chính xác.

* Yield (lợi suất) thực chất phản ánh kỳ vọng của thị trường. Khi thời gian trôi đi, nếu lãi suất thị trường không đổi, giá của Premium Bond sẽ giảm dần về mệnh giá, còn giá của Discount Bond sẽ tăng dần về mệnh giá.
* Đối với trái phiếu thông thường, điểm đến cuối cùng của giá là Mệnh giá (Par Value) tại ngày đáo hạn (YTM).
* Nhưng đối với Callable Bond, nếu thị trường thuận lợi cho doanh nghiệp, điểm đến của giá sẽ bị "bẻ lái", tiến dần về Giá mua lại (Call Price) tại ngày bị gọi (YTC) chứ không đi hết con đường đến ngày đáo hạn nữa.

2. Về việc đánh giá "Doanh nghiệp có Call Bond hay không?"
Bạn nói hoàn toàn ĐÚNG! Việc đánh giá doanh nghiệp có call bond hay không tại thời điểm $t$ chẳng liên quan gì đến việc nhìn vào con số YTC cả.
Bởi vì:

* YTC chỉ là một con số "Giả định - Kịch bản nếu": Đúng như bạn nói, nó trả lời cho câu hỏi: "Nếu tôi mua hôm nay và ông gọi lại vào ngày đó, thì lãi suất tăng trưởng hàng năm (lợi nhuận) của tôi là bao nhiêu?". Nó là kết quả tính toán sau khi có giá thị trường, chứ nó không phải là động lực thúc đẩy doanh nghiệp.
* Động lực thực sự của doanh nghiệp: Họ hoàn toàn không quan tâm YTC của bạn là bao nhiêu. Họ chỉ quan tâm đến túi tiền của họ. Họ sẽ lấy Coupon Rate (cố định trên tờ bond) đem so sánh trực tiếp với Lãi suất đi vay hiện tại của thị trường (Market Interest Rate) tại thời điểm $t$ đó.

👉 Quy luật bất biến:

* Nếu $\text{Lãi suất thị trường} < \text{Coupon Rate}$ $\rightarrow$ Doanh nghiệp sẽ Call Bond (để vay chỗ mới rẻ hơn, trả nợ chỗ cũ đắt hơn).
* Nếu $\text{Lãi suất thị trường} \ge \text{Coupon Rate}$ $\rightarrow$ Doanh nghiệp KHÔNG Call Bond (chẳng dại gì đi vay mới với lãi suất cao hơn để trả nợ cho khoản vay cũ đang có lãi suất rẻ).

------------------------------

💡 Mối liên kết cuối cùng để bạn thông suốt hoàn toàn:

Tại sao trong các bài tập (như bài trước bạn vừa làm), người ta lại kết luận: "Vì YTC < Coupon Rate nên doanh nghiệp sẽ Call"?
Đó thực chất là một cách nói tắt (đôi khi làm học viên bị rối bản chất).

Như tôi đã giải thích ở câu trước, trên thị trường tài chính, khi lãi suất thị trường giảm xuống, các nhà đầu tư sẽ tranh nhau mua trái phiếu đó khiến giá trái phiếu tăng lên. Khi giá tăng đến mức bão hòa, con số YTC tính ra sẽ tự động bằng đúng với Lãi suất thị trường.


Cho nên, việc họ so sánh YTC < Coupon Rate thực chất chỉ là một hệ quả gián tiếp của việc Lãi suất thị trường < Coupon Rate mà thôi! Tư duy đi từ bản chất gốc của bạn (so sánh Lãi suất thị trường với Coupon Rate) mới là tư duy của một nhà quản trị tài chính thực thụ.
