# Lecture 9-10 Recap

---

## 1. Outline

Before derivatives, you have studied:
- Risk and return
- CAPM and APT
- Portfolio diversification
- Systematic vs idiosyncratic risk

**Derivatives extend these ideas** by allowing investors and firms to:
- Transfer specific risks
- Lock in prices or rates
- Create asymmetric payoff structures
- Isolate exposures without trading the underlying asset

> **Key message:**  
> Derivatives do not eliminate risk — they **reallocate and reshape** it.

---

## 2. What Is a Derivative?

A **derivative** is a financial contract whose value depends on an **underlying**, such as:
- Equity prices or indices
- Interest rates
- Exchange rates
- Commodities
- Credit events

### Classification Framework (Very Important)

| Dimension | Categories |
|--------|-----------|
| Instrument | Forwards, Futures, Options, Swaps |
| Trading Venue | Exchange-traded vs OTC |
| Payoff Shape | Linear vs Non-linear |

- **Linear payoffs:** forwards, futures, swaps  
- **Non-linear payoffs:** options  

---

## 3. Forward & Futures Contracts

### Economic Meaning
A forward or futures contract **locks in a price today** for a transaction at a future date.

- **Long position:** obligation to buy
- **Short position:** obligation to sell

### Forward Payoff


For a **long forward**:
$
\Pi_T = S_T - F_0
$

**Key properties**
- Linear payoff
- Zero value at initiation
- Symmetric gains and losses

---

### Futures vs Forwards (High-Yield Exam Topic)

| Feature | Forward | Futures |
|------|--------|--------|
| Market | OTC | Exchange-traded |
| Settlement | At maturity | Daily (mark-to-market) |
| Counterparty risk | High | Low |
| Liquidity | Low | High |

> **Core intuition:**  
> Daily marking-to-market significantly reduces counterparty risk in futures.

---

## 4. No-Arbitrage Pricing Logic

### Law of One Price
> Two portfolios with identical future payoffs must have the same price today.

### Forward Price (No Income Asset)
$
F_0 = S_0 e^{rT}
$

If this condition does not hold:
- Arbitrageurs buy the undervalued side
- Sell the overvalued side
- Prices adjust until no arbitrage remains


::contentReference[oaicite:1]{index=1}


**Important:**  
No-arbitrage ensures **price consistency**, not guaranteed profits for all investors.

---

## 5. Options: Asymmetric Payoffs

Options differ fundamentally from forwards and futures:
- The buyer has a **right, not an obligation**
- The buyer pays a **premium**
- The seller receives the premium and bears risk

### Option Payoff Diagrams

::contentReference[oaicite:2]{index=2}


- **Call option payoff:** $\max(S_T - K, 0)$
- **Put option payoff:** $\max(K - S_T, 0)$

---

### Risk Profiles (Frequently Tested)

| Position | Maximum Loss | Maximum Gain |
|-------|-------------|-------------|
| Long Call | Premium | Unlimited |
| Long Put | Premium | Limited |
| Short Call | Unlimited | Premium |
| Short Put | Large | Premium |

> Think in terms of **insurance**:  
> Buyers buy protection, sellers earn premiums for bearing risk.

---

## 6. Option Value Decomposition

$
\text{Option Value} = \text{Intrinsic Value} + \text{Time Value}
$

- **Intrinsic value:** value if exercised today
- **Time value:** value of future uncertainty

### Effect of Volatility


- Higher volatility → higher option value
- Volatility benefits **option holders**
- Volatility is itself a priced risk

---

## 7. Hedging vs Speculation vs Arbitrage

### Hedging
- Objective: reduce risk
- Example: airline hedging fuel prices
- Instruments: futures, options, swaps

### Speculation
- Objective: profit from price movements
- Willingly takes on risk

### Arbitrage
- Objective: riskless profit
- Exploits mispricing
- Enforces no-arbitrage conditions

> The **same derivative** can serve different purposes depending on intent.

---

## 8. Swaps (Conceptual Bridge)

### Interest Rate Swap

- Exchange fixed-rate payments for floating-rate payments (or vice versa)
- Used to alter interest-rate exposure
- Does not change the underlying loan principal

**Key insight:**  
A swap can be replicated by a **portfolio of bonds**.

---

## 9. Common Student Pitfalls

1. Confusing **payoff** with **profit**
2. Forgetting to subtract the option premium
3. Mixing up futures and forward settlement
4. Believing derivatives create risk
5. Ignoring the purpose (hedging vs speculation)

---

## 10. Exam-Oriented Summary

### Must-Know Relationships
- Futures vs forwards → settlement mechanism
- Options → asymmetric payoffs
- Arbitrage → law of one price
- Volatility → increases option value

### One-Sentence Takeaways
- Forwards lock in prices  
- Futures reduce counterparty risk  
- Options reshape risk profiles  
- Swaps transform exposures  
- Arbitrage keeps markets coherent
