## 1. Empirical Failures of CAPM

CAPM pricing equation:

$$
E(R_i) = R_f + \beta_i \big(E(R_M) - R_f \big)
$$

where  
- $\beta_i = \dfrac{\operatorname{Cov}(R_i, R_M)}{\operatorname{Var}(R_M)}$

Empirical evidence for **Systematic failures**:

- **Small-Firm Effect (SMB):** Small-cap stocks have **$\alpha_i > 0$** vs. CAPM.
- **Value Effect (HML):** High Book-to-Market (Value) stocks show **positive alphas**.
- Many portfolios earn returns **above** their CAPM-implied expected returns.

  
CAPM misprices assets because *market beta alone* cannot capture all relevant risk factors.

The **alpha** of asset $i$ relative to CAPM is:

$$
\alpha_i = R_i - \Big[ R_f + \beta_i \big(R_M - R_f \big) \Big]
$$

---

## Where Stocks Lie On The SML

**Why CAPM fails here:**  
* When multiple risks are present, CAPM (with only $\beta_{iM}$) produces pricing errors (alphas) because it ignores non-market factors.
* Many returns are found above the SML.

## 2. Intertemporal CAPM (ICAPM – Merton, 1973)
* Investors care about today's investment returns as well as how today's decisions affect future expected returns.
* ICAPM extends CAPM into a multi-period world.  
* The basic pricing relation for excess returns is:

$$
E(R_{i} - R_f) = \beta_{iM} \lambda_M + \sum_{k=1}^{K} \beta_{ik} \lambda_k
$$

where  
- $\beta_{iM}$ = sensitivity of asset $i$ to the market portfolio  
- $\beta_{ik}$ = sensitivity to state variable $k$ (e.g., interest rates, volatility, GDP growth)  
- $\lambda_M$ = market risk premium  
- $\lambda_k$ = risk premium for state variable $k$

**Key ideas:**

- Investors care about **current returns** and **future investment opportunities**.
- State variables include **interest rates, volatility, inflation, GDP growth**, etc.
- Assets that hedge against worsening opportunities (negative $\beta_{ik}$ to “bad” states) earn **lower risk premia**.

## 3. Arbitrage Pricing Theory (APT, 1976)
* If arbitrage opportunities exist, rational traders will eliminate them.
* Therefore, asset prices must adjust so that no risk-free profit can be made from trading portfolios.

***What must expected returns look like if there are no free arbitrage opportunities?***
* APT **assumes** that expected excess returns are **linear in factor loadings**:

$$
E(R_i) = R_f + \beta_{i1} \lambda_1 + \beta_{i2} \lambda_2 + \cdots + \beta_{iK} \lambda_K
$$

**Realized** excess returns can be written as:

$$
R_i - R_f = \beta_{i1} F_1 + \beta_{i2} F_2 + \cdots + \beta_{iK} F_K + \epsilon_i
$$

where  
- $F_k$ = realization of factor $k$ (mean $\lambda_k$)  
- $\beta_{ik}$ = sensitivity of asset $i$ to factor $k$  
- $\epsilon_i$ = idiosyncratic (diversifiable) error

If an asset is mispriced, we observe a **non-zero alpha**:

$$
\alpha_i = E(R_i) - \Big[ R_f + \sum_{k=1}^{K} \beta_{ik} \lambda_k \Big] \neq 0
$$

Arbitrageurs can construct **zero-net-investment portfolios** with:

- Zero factor exposure (cancel $\beta_{ik}$), but  
- Positive expected return (from $\alpha_i > 0$),

Arbitrageurs will buy undervalued assets and short overvalued ones, pushing prices until:

$$
\alpha_i → 0
$$

***In the absence of arbitrage, expected returns must be linear in factor loadings, and persistent alpha cannot exist.***

## 4. Fama–French 3-Factor Model

***Fama and French asked: "Which factors actually explain the anomalies we observe?"***
The FF3 model specifies excess returns as:

$$
R_i - R_f = \alpha_i + \beta_{iM} (R_M - R_f) + s_i \, \text{SMB} + h_i \, \text{HML} + \epsilon_i
$$

where  
- $R_i$ = return on asset/portfolio $i$  
- $R_M$ = market return  
- $\text{SMB}$ = Small Minus Big (size factor)  
- $\text{HML}$ = High Minus Low (value factor)  
- $\beta_{iM}$, $s_i$, $h_i$ = factor loadings  
- $\alpha_i$ = pricing error relative to FF3  
- $\epsilon_i$ = idiosyncratic error

**Interpretation:**

- CAPM: only $(R_M - R_f)$ explains excess returns.  
- FF3: adds **size** and **value** factors to capture small-firm and value premia.
- **Note**: In many tests, $\alpha_i$ becomes statistically insignificant once SMB and HML are included.

## 5. Limits of Arbitrage & Behavioral Finance

Even if *risk-based* models are correct, prices can deviate from fundamentals because of:

### 5.1 Taste-Based Preferences (ESG Example)

Suppose investor utility depends on both **wealth** and a **“green” preference index** $G$:

$$
U = U(W_T, G)
$$

where  
- $W_T$ = terminal wealth  
- $G$ = share of green assets in the portfolio

An investor may accept a **lower expected return** for green portfolios:

$$
E(R_{\text{Green}}) < E(R_{\text{Brown}})
\quad \text{even if} \quad
\text{risk}(\text{Green}) \approx \text{risk}(\text{Brown})
$$

### 5.2 Efficiency Loss (Sharpe Ratio)

Sharpe ratio of a portfolio $p$:

$$
\text{SR}_p = \frac{E(R_p) - R_f}{\sigma_p}
$$

If investors exclude “Brown” assets, the **constrained green portfolio** often has:

$$
\text{SR}_{\text{Green-only}} < \text{SR}_{\text{Tangency}}
$$

showing an **efficiency cost** of non-financial preferences.


Behavioral biases and taste-based investing can sustain **pricing anomalies** even when **arbitrage is conceptually possible.**

### Solution 1:
**a)**  Theoretical expected return of portfolio:
* **APT says:**\
$
E^{APT}(R_P) - R_f = \beta_{P,GDP}*\lambda_{GDP} + \beta_{P,INF}*\lambda_{INF}
               + \beta_{P,GOLD}*\lambda_{GOLD} + \beta_{P,MKT}*\lambda_{MKT}\
$
\
$
E^{APT}(R_P) - R_f = 0.8(3\%) + 0.6(1.3\%) + (-0.5)(0.8\%) + 1.4(5\%) \\ = 2.40\% + 0.78\% - 0.40\% + 7.00\% \\ = 9.78\% \
$

* Now add the risk-free rate $R_f = 1\%$:\
$
E^{APT}(R_P) = R_f + 9.78\% = 1\% + 9.78\% = \boxed{10.78\%}
$


**b)**
$
\alpha_{P} = E(R_P) - E^{APT}(R_P) = 13\% - 10.78\% = \boxed{2.22\%}
$
➜ The portfolio earns 2.22% more than what APT says it “should” for its factor loadings.
From an arbitrage / active management perspective:

* α > 0 → cheap (undervalued) or earning a higher return for its risk.\
➜ Go LONG this portfolio and SHORT the matching factor hedge.

Choose weights so factor exposures cancel → pure alpha.

### Solution 2:

**a)** Calculate the Hedge Weight ($w_H$)
**Goal:** Make the Net Oil Beta equal to zero.

The equation for the net exposure is:
$1 \cdot \beta_{P, Oil} + w_H \cdot \beta_{H, Oil} = 0$

Substitute the values:
$1 \cdot (0.6) + w_H \cdot (-1.2) = 0$
$0.6 - 1.2w_H = 0$
$1.2w_H = 0.6$
$w_H = \frac{0.6}{1.2}$
$\mathbf{w_H = \boxed{0.5}}$

**Answer:** You must add **$0.50** of the Energy Hedge Fund for every $1.00 of your existing portfolio.

**b)** Calculate the New Market Beta
**Goal:** Find the weighted average Market Beta of the new total position.

* **Initial Value:** $1.0$
* **Added Value:** $0.5$
* **Total New Value:** $1.5$

The formula for the new Market Beta is:
$\beta_{New, Mkt} = \frac{(1 \cdot \beta_{P, Mkt}) + (w_H \cdot \beta_{H, Mkt})}{1 + w_H}$

Substitute the values:
$\beta_{New, Mkt} = \frac{(1 \cdot 1.2) + (0.5 \cdot 0.3)}{1.5}$
$\beta_{New, Mkt} = \frac{1.2 + 0.15}{1.5}$
$\beta_{New, Mkt} = \frac{1.35}{1.5}$
$\mathbf{\beta_{New, Mkt} = 0.9}$

The new Market Beta is **0.9**.

### Solution 3:

**a)** **Implied Factor Risk Premium**

From Portfolio D:
$
\lambda = \frac{E[r_D] - r_f}{\beta_D}
= \frac{16\% - 4\%}{1.5}
= 8\%
$

From Portfolio E:
$
\lambda = \frac{8\% - 4\%}{0.5}
= 8\%
$

$
\boxed{\lambda = 8\%}
$



**b)** **Mispricing Check**

Using the APT pricing equation:

- Portfolio D:
$
\hat E[r_D] = 4\% + 1.5(8\%) = 16\%
$

- Portfolio E:
$
\hat E[r_E] = 4\% + 0.5(8\%) = 8\%
$

Observed and theoretical returns coincide.

$
\boxed{\text{No arbitrage opportunity exists}}
$



**c)** **Zero-Investment, Zero-Beta Arbitrage Portfolio (Hypothetical)**

To replicate the beta of Portfolio D using E and F:

$
0.5w = 1.5 \quad \Rightarrow \quad w = 3
$

Replicating portfolio:
$
3E - 2F
$

Expected return:
$
E[r_{3E - 2F}] = 3(8\%) - 2(4\%) = 16\%
$

Construct the arbitrage position:
$
\Pi = D - (3E - 2F)
$

Beta:
$
\beta_\Pi = 1.5 - (3 \times 0.5 - 2 \times 0) = 0
$

Initial cost:
$
0
$


**d)** **Expected Arbitrage Profit**

$
E[r_\Pi] = E[r_D] - E[r_{3E - 2F}] = 16\% - 16\% = 0\%
$

$
\boxed{\text{No arbitrage profit under the given prices}}
$

> **Note:** If Portfolio D’s expected return were higher (lower) than 16%, the strategy above would generate a positive arbitrage profit by going long (short) Portfolio D and short (long) the replicating portfolio.

### Solution 4:

**a)** Benchmarked Expected Total Return

First compute the fund’s **benchmarked expected excess return**:

$
E[R^{excess}_{Fund}] = 0.9(6.5\%) + 0.6(2.5\%) - 0.3(4.0\%)
$

Compute each term:

- Market contribution: \(\;0.9 \times 6.5\% = 5.85\%\)
- SMB contribution: \(\;0.6 \times 2.5\% = 1.50\%\)
- HML contribution: \(\;-0.3 \times 4.0\% = -1.20\%\)

So,

$
E[R^{excess}_{Fund}] = 5.85\% + 1.50\% - 1.20\% = 6.15\%
$

Convert to **expected total return** by adding the risk-free rate:

$
E[r_{Fund}] = r_f + E[R^{excess}_{Fund}] = 3.0\% + 6.15\% = 9.15\%
$

$
\boxed{E[r_{Fund}] = 9.15\%}
$


**b)** Fund’s Alpha Given Actual Total Return = 11.2%

Actual **excess** return:

$
R^{excess}_{actual} = 11.2\% - 3.0\% = 8.2\%
$

Alpha is actual excess return minus benchmarked excess return:

$
\alpha = 8.2\% - 6.15\% = 2.05\%
$

$
\boxed{\alpha = 2.05\% \text{ (annual)}}
$


**c)** Interpretation of Factor Loadings (Style)

- $\beta_M = 0.9$: slightly below-market market exposure (near-market beta)
- $\beta_{SMB} = +0.6$: **positive size tilt** → **small-cap** exposure
- $\beta_{HML} = -0.3$: **negative value tilt** → **growth** exposure (anti-value)

$
\boxed{\text{Style: Small-cap Growth (small tilt + growth tilt), with near-market exposure.}}
$
